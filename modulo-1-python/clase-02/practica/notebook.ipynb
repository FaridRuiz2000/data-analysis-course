### Ejercicio 1 — Clasificar una transacción

En cualquier sistema de pagos existe una lógica de clasificación automática. Dependiendo del monto, una transacción recibe una categoría distinta que determina cómo se procesa.

**Consigna:** Dado el monto de una transacción, escribí un programa que lo clasifique según estas reglas y lo imprima con un mensaje descriptivo:


- Menos de $1.000: "micro transacción"
- Entre $1.000 y $9.999: "transacción estándar"
- Entre $10.000 y $99.999: "transacción importante"
- $100.000 o más: "transacción de alto valor"

```python
# Punto de partida
monto = 45750

if monto > 0 and monto < 1000:
	print("micro transaccion")
elif monto >= 1000 and monto <= 9999:
	print("transaccion standar")
elif monto >= 10000 and monto <= 99999:
	print("transaccion importante")
else:
	print("transaccion de alto valor")
# Tu tarea: clasificar el monto e imprimir el resultado
# Ejemplo de salida: "Monto: $45750 — Categoría: transacción importante"
```

### Ejercicio 2 — Recorrer una lista de precios

Una tienda necesita calcular automáticamente el precio final de todos sus productos aplicando el impuesto correspondiente.

**Consigna:** Recorre la lista de precios con un bucle for y para cada precio calculá e imprimí el valor sin impuesto y con un impuesto del 19%. Al final, imprimí también el total acumulado de todos los precios con impuesto.

```python
# Punto de partida
precios = [8500, 32000, 15900, 7200, 54000, 12300]
impuesto = 0.19
total = 0

# Tu tarea: recorrer la lista, calcular el precio final de cada producto
# y acumular el total general
# Ejemplo de salida por producto: "Precio base: $8500 → Precio final: $10115.0"
# Al final: "Total general: $..."

for precio in precios:
  precio_final = precio * (1 + impuesto)
  print(f"Precio base: ${precio} → Precio final: ${precio_final}")

  total += precio_final

print(f"Total general: ${total}")
```

### Ejercicio 3 — Buscar un registro

En un sistema de inventario necesitas verificar si un producto existe antes de intentar consultarlo. Acceder a una clave que no existe genera un error que rompe el programa.

**Consigna:** Dado el diccionario de productos y un código de búsqueda, verificá si el producto existe e imprimí su información. Si no existe, imprimí un mensaje indicándolo. Usá .get() para el acceso seguro.

```python
# Punto de partida
inventario = {
    "NB-001": {"nombre": "Notebook Pro", "precio": 320000, "stock": 5},
    "MS-002": {"nombre": "Mouse Inalámbrico", "precio": 8500, "stock": 42},
    "MN-003": {"nombre": "Monitor 24\"", "precio": 210000, "stock": 0},
    "TC-004": {"nombre": "Teclado Mecánico", "precio": 45000, "stock": 12},
}

codigo_busqueda = "MN-003"

# Tu tarea: buscar el código en el inventario
# Si existe: imprimir nombre, precio y stock
# Si no existe: imprimir "Producto no encontrado"
if inventario.get(codigo_busqueda) is not None:
  for clave, valor in inventario[codigo_busqueda].items():
    print(f"{clave}: {valor}")
else:
  print("Producto no encontrado")
```

### Ejercicio 4 — Filtrar con condiciones

Un analista necesita separar los registros de ventas en dos grupos: las que superan la meta mensual y las que no. Es el tipo de operación de filtrado más básica en data analysis.

**Consigna:** Recorre la lista de ventas y separalas en dos listas nuevas: sobre_meta y bajo_meta. Al final imprimí cuántas hay en cada grupo y el total acumulado de cada uno. La meta es $50.000.

```python
# Punto de partida
ventas = [23000, 87000, 45000, 102000, 31500, 68000, 49000, 95000, 12000, 74000]
meta = 50000

sobre_meta = []
bajo_meta = []

# Tu tarea: clasificar cada venta en el grupo correcto y calcular los totales
for venta in ventas:
  if venta > meta:
    sobre_meta.append(venta)
  else:
    bajo_meta.append(venta)

total_sobre_meta = sum(sobre_meta)
total_bajo_meta = sum(bajo_meta)

print(f"Total de ventas que superan la meta mensual: ${total_sobre_meta} \nTotal de ventas que no superan la meta mensual: ${total_bajo_meta}")
```

### Ejercicio 5 — Reporte por categoría

En un reporte de ventas real necesitas agrupar los datos por algún criterio — categoría, región, vendedor — y calcular métricas por grupo. Antes de usar pandas, este ejercicio te muestra cómo funciona esa lógica por dentro.

**Consigna:** Recorre la lista de productos y construí un diccionario que agrupe el total de ventas (precio × cantidad) por categoría. Al final imprimí un reporte con el total por categoría y cuál fue la categoría con mayor ingreso.

```python
# Punto de partida
productos = [
    {"nombre": "Notebook",    "categoria": "tecnologia", "precio": 320000, "cantidad": 3},
    {"nombre": "Escritorio",  "categoria": "muebles",    "precio": 85000,  "cantidad": 5},
    {"nombre": "Mouse",       "categoria": "tecnologia", "precio": 8500,   "cantidad": 20},
    {"nombre": "Silla",       "categoria": "muebles",    "precio": 45000,  "cantidad": 8},
    {"nombre": "Monitor",     "categoria": "tecnologia", "precio": 210000, "cantidad": 4},
    {"nombre": "Lámpara",     "categoria": "muebles",    "precio": 12000,  "cantidad": 15},
    {"nombre": "Auriculares", "categoria": "tecnologia", "precio": 32000,  "cantidad": 11},
]

totales_por_categoria = {}

# Tu tarea: construir el diccionario de totales y encontrar la categoría ganadora
for producto in productos:
  categoria = producto["categoria"]

  if categoria not in totales_por_categoria:
    totales_por_categoria[categoria] = 0
    
  totales_por_categoria[categoria] += producto["precio"] * producto["cantidad"]

categoria_ganadora = ""
total_categoria_ganadora = 0

for categoria, total in totales_por_categoria.items():
  if total > total_categoria_ganadora:
    categoria_ganadora = categoria
    total_categoria_ganadora = total

  print(f"{categoria.capitalize()} - valor total: ${total:,.2f}")

print(f"Categoria con mayor ingreso: {categoria_ganadora.capitalize()} - valor total: ${total_categoria_ganadora:,.2f}")
```

### Ejercicio 6 — Validar y limpiar datos

En la práctica, los datos que llegan de fuentes externas rara vez están limpios. Valores negativos, campos vacíos, tipos incorrectos — todo eso aparece antes de poder analizar nada. Saber detectarlo y manejarlo es una habilidad central en data analysis.

**Consigna:** Recorre la lista de registros e identificá los que tienen problemas. Un registro es inválido si el precio es negativo, si el nombre está vacío, o si el stock no es un número entero. Separalos en dos listas: validos e invalidos, y para cada inválido indicá el motivo.

```python
# Punto de partida
registros = [
    {"nombre": "Notebook",   "precio": 320000, "stock": 5},
    {"nombre": "",           "precio": 8500,   "stock": 10},
    {"nombre": "Monitor",    "precio": -210000,"stock": 2},
    {"nombre": "Teclado",    "precio": 45000,  "stock": 3.5},
    {"nombre": "Mouse",      "precio": 8500,   "stock": 0},
    {"nombre": "Auriculares","precio": 32000,  "stock": 7},
    {"nombre": "Cámara",     "precio": 0,      "stock": 4},
]

validos = []
invalidos = []

# Tu tarea: clasificar cada registro y registrar el motivo de rechazo
# Ejemplo de salida para inválidos:
# {"nombre": "", "precio": 8500, "stock": 10, "motivo": "nombre vacío"}
for producto in registros:
  try:
    if producto["nombre"] == "":
      raise ValueError("Nombre vacio")
    if producto["precio"] <= 0:
      raise ValueError(f"Precio invalido: {producto['precio']}")
    if isinstance(producto["stock"], int) == False:
      raise ValueError("El stock debe ser un numero entero")

    valido = producto.copy()
    validos.append(valido)
  except ValueError as e:
    invalido = producto.copy()
    invalido["error"] = str(e)
    invalidos.append(invalido)

print(f"Registros validos: {validos}")
print(f"Registros invalidos: {invalidos}")
```

### Ejercicio 7 — Generador de reportes 

En un sistema real de ventas, los reportes se generan automáticamente a partir de los datos crudos. Este ejercicio simula exactamente eso: a partir de una lista de transacciones, construís un reporte completo que podría enviarse por email o guardarse en un archivo.

**Consigna:** A partir de la lista de transacciones, construye un reporte que incluya:


- Total de transacciones procesadas
- Monto total y promedio
- Desglose por país (cantidad y monto total)
- Las 3 transacciones de mayor monto
- Transacciones rechazadas (monto negativo o cero) con su motivo

```python
# Punto de partida
transacciones = [
    {"id": "TX001", "pais": "MX", "monto": 45000,  "moneda": "MXN"},
    {"id": "TX002", "pais": "AR", "monto": 128000, "moneda": "ARS"},
    {"id": "TX003", "pais": "CO", "monto": 0,      "moneda": "COP"},
    {"id": "TX004", "pais": "MX", "monto": 87000,  "moneda": "MXN"},
    {"id": "TX005", "pais": "AR", "monto": -5000,  "moneda": "ARS"},
    {"id": "TX006", "pais": "PE", "monto": 32000,  "moneda": "PEN"},
    {"id": "TX007", "pais": "CO", "monto": 215000, "moneda": "COP"},
    {"id": "TX008", "pais": "MX", "monto": 61000,  "moneda": "MXN"},
    {"id": "TX009", "pais": "AR", "monto": 94000,  "moneda": "ARS"},
    {"id": "TX010", "pais": "PE", "monto": 17000,  "moneda": "PEN"},
]

# Tu tarea: construir el reporte completo
# Pista: para las top 3 investigá sorted() con el parámetro key=
transacciones_procesadas = len(transacciones)

transacciones_validas = []
transacciones_invalidas = []

for transaccion in transacciones:
  try:
    if transaccion["monto"] <= 0:
      raise ValueError(f"Monto invalido: {transaccion['monto']}")

    valido = transaccion.copy()
    transacciones_validas.append(valido)
  except ValueError as e:
    invalido = transaccion.copy()
    invalido["error"] = str(e)
    transacciones_invalidas.append(invalido)

monto_total = 0
cantidad_transacciones = len(transacciones_validas)

for transaccion in transacciones_validas:
  monto_total += transaccion["monto"]

promedio = monto_total / cantidad_transacciones

agrupados = {}
for registro in transacciones_validas:
  pais = registro["pais"]

  if pais not in agrupados:
    agrupados[pais] = []

  agrupados[pais].append(registro)

ordenadas = sorted(transacciones_validas, key=lambda t: t["monto"], reverse=True)
top_3 = ordenadas[:3]

print("=== REPORTE DE TRANSACCIONES ===")

print(f"Total de transacciones procesadas: {transacciones_procesadas}")
print(f"Monto total: ${monto_total:,.2f}")
print(f"Monto promedio: ${promedio:,.2f}")

print("\n--- Desglose por país ---")
for pais, registros in agrupados.items():
    cantidad = len(registros)
    monto = sum(registro["monto"] for registro in registros)
    print(f"{pais} - Cantidad: {cantidad} | Monto total: ${monto:,.2f}")

print("\n--- Top 3 transacciones ---")
for transaccion in top_3:
    print(f"{transaccion['id']} - {transaccion['pais']} - ${transaccion['monto']:,.2f}")

print("\n--- Transacciones rechazadas ---")
for transaccion in transacciones_invalidas:
    print(f"{transaccion['id']} - Monto: ${transaccion['monto']:,.2f} - Motivo: {transaccion['error']}")
```
